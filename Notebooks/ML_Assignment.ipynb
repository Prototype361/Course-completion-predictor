# Day 2 - Data Exploration Findings

## Data Quality Issues Discovered

During exploration of the training dataset, several data quality issues were identified.

### 1. Missing Values
Several useful learner features contain missing values. The highest missingness among potentially useful features is in `age` (152 rows, 6.12%), followed by `video_watch_pct` (147 rows, 5.92%), `hours_per_week_pledged` (108 rows, 4.35%), `weekly_logins_avg` (96 rows, 3.86%), `education` (90 rows, 3.62%), `has_mentor` (64 rows, 2.58%), and `city_tier` (53 rows, 2.13%).

There are also 9 rows where `completed_course`, the target variable, is missing. These rows cannot provide a known outcome for supervised training.

### 2. Duplicate Records
84 rows contain learner IDs that appear more than once. Of these, 83 are exact duplicate rows. One learner, `L-01467`, appears twice with identical feature information, but one record has a missing `completed_course` value while the other has `completed_course = 1`.

The duplicate records are consistent with the export issue described in the Data Dictionary.

### 3. Invalid Age Values
The `age` column contains impossible values. Six learners have an age of `-3`, while 16 learners have ages of `199` or `300`. These values are inconsistent with realistic learner ages and will need to be handled during cleaning.

### 4. Impossible Weekly Hours
The `hours_per_week_pledged` column contains 9 values greater than 168 hours per week. Since a week contains only 168 hours, these values are physically impossible and indicate invalid input data.

### 5. Unexpected Login-Day Values
The `days_since_last_login` column contains 987 values greater than 21. According to the Data Dictionary, this value measures the number of days since the learner's last login during the first three weeks, measured at the end of week 3. Values greater than 21 therefore require investigation.

### 6. Video Watch Percentages Above 100%
38 learners have `video_watch_pct` values above 100%, with the maximum being 115.3%. The Data Dictionary explains that the video player can over-report this value when learners rewatch content. These values are therefore suspicious but may reflect known system behaviour rather than simple data-entry errors.

### 7. Inconsistent Education Categories
The `education` column contains multiple representations of apparently identical categories, such as `Bachelors`, `bachelors`, `Bachelor's`, and `bachelor degree`. Similar inconsistencies exist for high school, diploma, masters, and other categories. These labels will need to be standardized before modelling.

### 8. Inconsistent Mentor Values
The `has_mentor` column contains multiple representations of the same boolean values, including `0`, `No`, `FALSE`, `no`, `N`, `1`, `TRUE`, `Yes`, `Y`, and `yes`. These will need to be converted into a consistent representation.

### 9. Inconsistent Device Categories
The `device` column contains inconsistent labels and capitalization, including `mobile`, `Mobile`, `MOBILE`, `phone`, `laptop`, `Laptop`, `notebook`, `desktop`, `Desktop`, `PC`, `tablet`, `Tablet`, and `ipad`. These categories will need to be standardized.

### 10. Inconsistent City Tier Categories
The `city_tier` column contains numeric and text representations of the same categories. For example, Tier 2 appears as `2`, `T2`, `Tier 2`, and `tier-2`. Similar inconsistencies exist for Tiers 1 and 3. There are also 53 missing values.

### 11. Inconsistent Signup Date Formats
The `signup_date` column is stored as an object/text column and contains multiple date formats, including `YYYY-MM-DD`, `DD-MM-YYYY`, and `DD/MM/YYYY`. The column must be converted into a consistent datetime representation before reliable date-based analysis can be performed.

### 12. Future Information / Data Leakage
Two columns contain information that would not be available when the prediction is supposed to be made. `final_score` is calculated at the end of the 12-week program, and `certificate_issued` is recorded after the program ends. Since predictions are supposed to be made at the end of week 3, these columns must not be used as model inputs.

The `notes` column is also highly incomplete and unstructured, with approximately half of its values missing. It will likely be excluded from the initial model.

### Duplicate Records

The raw training dataset contained 83 exact duplicate rows. These rows were removed because they contained no additional information and would cause duplicated learners to receive extra weight during model training.

After removal, no exact duplicate rows remain.

```python
clean_df[clean_df["learner_id"].duplicated(keep=False)]
```

```python
clean_df = clean_df[
    ~(
        (clean_df["learner_id"] == "L-01467") &
        (clean_df["completed_course"].isna())
    )
]
```

```python
clean_df[clean_df["learner_id"] == "L-01467"]
```

```python
clean_df["learner_id"].duplicated().sum()
```

### Duplicate Learner ID with Missing Outcome

After removing exact duplicate rows, learner `L-01467` still appeared twice. The two records had identical feature values, but one had `completed_course = 1` while the other had a missing completion outcome.

The duplicate record with the missing target was removed because an identical record containing the known outcome was already present.

```python
clean_df["completed_course"].isna().sum()
```

```python
clean_df = clean_df.dropna(subset=["completed_course"])
```

```python
clean_df["completed_course"].isna().sum()
```

### Missing Target Values

The training data originally contained 9 rows with a missing `completed_course` value. These rows cannot be used for supervised learning because the correct outcome is unknown.

One of these rows was the duplicate `L-01467` record already removed during duplicate handling. The remaining 8 rows were removed from the cleaned training dataset.

### Invalid Age Values

The `age` column contained 22 clearly invalid values: 6 values of `-3` and 16 values of `199` or `300`.

Instead of removing the entire learner records, these invalid age values were replaced with missing values (`NaN`). This preserves the other useful information from those learners while allowing the invalid age values to be handled together with the existing missing ages later.

```python
clean_df[clean_df["hours_per_week_pledged"] > 168][
    ["learner_id", "hours_per_week_pledged"]
]
```

```python
clean_df.loc[
    clean_df["hours_per_week_pledged"] > 168,
    "hours_per_week_pledged"
] = np.nan
```
